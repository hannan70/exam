### Problem 1: Given a string s, reverse the order of its characters and return the reversed string.

```python
s = "hello"
revstr = "" 
for cha in s:  
    revstr = cha + revstr

print(revstr)
```

```python
s = "hello"
for i in range(len(s)-1, -1, -1):
    print(s[i], end="")
```

#### Problem 2: Given a positive integer n, calculate the sum of all even numbers from 1 to n, using while loop.


```python
n = 6
sum = 0
for i in range(1, n+1):
    if i % 2 == 0:
        sum += i

print(sum)
```

```python
# solve another way
n = 6   
i = 1
sum = 0
while i <= n:
    if i % 2 == 0:
        sum += i
    i += 1
print(sum)
```

Problem 3: Given a non-negative integer n, compute its factorial using recursion. The
factorial of n is the product of all positive integers less than or equal to n.

```python
def calculate_fac(n):
    if n == 0 or n == 1:
        return 1

    return n * calculate_fac(n - 1)

result = calculate_fac(5)
print(result)
```

#### Problem: Given an array containing only 0s, 1s, and 2s, sort the array in ascending order.


```python
# Way one
nums = [2, 0, 2, 1, 1, 0]
nums.sort()
print(nums)
```

```python
# another way
def sort_num(nums):
    n = len(nums)
    for i in range(n):
        for j in range(0, n - i - 1):
            if nums[j] > nums[j + 1]:
                # swap two numbers
                nums[j], nums[j + 1] = nums[j + 1], nums[j]

    return nums

nums = [2, 0, 2, 1, 1, 0]
result = sort_num(nums)
print(result)
```

#### Problem: Given two sorted arrays nums1 and nums2, merge them into a single sorted array.

```python
def merge_array(nums1, nums2):
    if 1 <= len(nums1) <= 100 and 1 <= len(nums2) <= 100:
        array = nums1 + nums2
        array.sort()
        return array

nums1 = [1, 3, 5] 
nums2 = [2, 4, 6]
result = merge_array(nums1, nums2)
print(result)
```
