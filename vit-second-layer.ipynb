```python
import torch
import torchvision.models as models
from torchvision import transforms
import matplotlib.pyplot as plt
from PIL import Image
```

```python
image_tensor = torch.rand(1, 3, 224, 224)
```

```python
model = models.vit_b_16(pretrained=True)
model.eval() 
```

```python
features_map = []

# callback function
def hook_fn(module, input, output):
    features_map.append(output)

# register the hook in 2nd layer
hook = model.encoder.layers[1].register_forward_hook(hook_fn)
```

```python
# forword pass for attention map capture
with torch.no_grad():
    _ = model(image_tensor)
```

```python
feature_vector = features_map[0]  
print(feature_vector.shape)
```

```python
# without batch dimention frint the final attention
attn = feature_vector.squeeze(0)
```
