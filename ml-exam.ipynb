
Problem: Implement a function to compute the gradients of the mean squared error
(MSE) loss with respect to the weights and bias of a linear regression model using PyTorch’s autograd. The model is defined as y = w · x + b, where w and b are learnable
parameters.


```python
import torch
```

```python
# input and true input
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=False)   
y_true = torch.tensor([2.0, 4.0, 6.0], requires_grad=False) 
```

```python
# make model 

 # random initial weight
w = torch.randn(1, requires_grad=True) 

# random initial bias
b = torch.randn(1, requires_grad=True)  

# make linear model
y_pred = w * x + b
```

```python
y_pred
```

```python
# calculat loss function
loss = ((y_pred - y_true) ** 2).mean()
```

```python
loss
```

```python
# calculate weight and bise
loss.backward()
# print the gredient
print(f"Gradient of w: {w.grad}")
print(f"Gradient of b: {b.grad}")
```

Problem: Approximate the Hessian matrix of a quadratic function f(w) = w
TAw +
b
Tw + c with respect to a weight vector w, where A is a symmetric matrix, b is a vector,
and c is a scalar. Use PyTorch to compute the Hessian via finite differences on the
gradients.

```python
import torch
```

```python
# take all input
A = torch.tensor([[2.0, 1.0], [1.0, 2.0]])
b = torch.tensor([1.0, 1.0])
c = 0.0
w = torch.tensor([1.0, 1.0], requires_grad=True)
```

```python
# quadratic function
def f(w):
    # n f(w) = wTAw +bTw + c 
    return w @ A @ w + b @ w + c

# Hessian Approximation Function
# epsilon is a small positive number 
# using epsilon for 
def approximate_hessian(f, w, epsilon=1e-4):
    # first dimension 0
    n = w.shape[0]
    # make 2d matrix using zero
    hessian = torch.zeros((n, n))

    for i in range(n):
        for j in range(n):
            # clone weight and detach the gradient
            w_ij1 = w.clone().detach()
            w_ij2 = w.clone().detach()
            w_ij3 = w.clone().detach()
            w_ij4 = w.clone().detach()
            
            # first point evaluate 
            w_ij1[i] += epsilon
            w_ij1[j] += epsilon
            f1 = f(w_ij1)

            # second point evaluate 
            w_ij2[i] += epsilon
            w_ij2[j] -= epsilon
            f2 = f(w_ij2)

            # third point evaluate 
            w_ij3[i] -= epsilon
            w_ij3[j] += epsilon
            f3 = f(w_ij3)

            # fourth point evaluate 
            w_ij4[i] -= epsilon
            w_ij4[j] -= epsilon
            f4 = f(w_ij4)

            # calculate Hessian matrix element
            hessian[i, j] = (f1 - f2 - f3 + f4) / (4 * epsilon ** 2)

    return hessian

# calculate hessian
hessian_approx = approximate_hessian(f, w)
print("Approximated Hessian matrix:\n", hessian_approx)
```

#### Extracting Attention Maps from ViT
Problem: Given a pretrained Vision Transformer (ViT) model from torchvision, extract
the attention maps from the last transformer layer for a given input image. The attention
maps represent the attention weights across patches.


```python
import torch
import torchvision.transforms as transforms
from torchvision.models import vit_b_16, ViT_B_16_Weights
from PIL import Image
import matplotlib.pyplot as plt
```

```python
# use vit_b_16 pretrained model for extracting the attention map
# Evaluation Mode on 
weights = ViT_B_16_Weights.IMAGENET1K_V1
model = vit_b_16(weights=weights)
model.eval()
```

```python
# vit model takes 224x224 resolution image
# random tensor (3x224x224) 
# 1 -> Batch size = 1 (single image)
# 3 → Number of channels = 3 (RGB image: Red, Green, Blue)
# 224 → Height = 224 pixels
# 224 → Width = 224 pixels
image_tensor = torch.rand(1, 3, 224, 224)  # (1, 3, 224, 224) = single batch
```

```python
# image processing


# ViT expects input normalized  When we input an image into a model it is very important to bring the pixel values of that image into a specific scale
normalize = transforms.Normalize(mean=(0.5, 0.5, 0.5), std=(0.5, 0.5, 0.5))
image_tensor = normalize(image_tensor[0]).unsqueeze(0)  # normalize and add batch dim
```

```python
attention_maps = []

def get_attention_hook(module, input, output):
    # add the last attention weights
    attention_maps.append(output.detach())

last_layer = model.encoder.layers[-1]
hook = last_layer.register_forward_hook(get_attention_hook)
```

#### Tracking NaN Loss in Neural Network Training

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
x = torch.randn(100, 10)  # 100 sample and 10 feature 
y = torch.randn(100, 1) # 100 sample 1 output
```

```python
x[:10]
```

```python
y[:10]
```

```python
# make 2 layer nural network
class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.layer1 = nn.Linear(10, 50)  # input size 10, hidden layer 50
        self.layer2 = nn.Linear(50, 1)   # hidden layer 50, output size 1

    def forward(self, x):
        x = torch.relu(self.layer1(x))  # use ReLU activation function
        x = self.layer2(x)
        return x
```

```python
# setup model loss and optimization
model = SimpleNN()  # create model
criterion = nn.MSELoss()  # MSE loss function
optimizer = optim.SGD(model.parameters(), lr=10.0)
```

```python
def train_model(x, y, model, criterion, optimizer, epochs=100):
    for epoch in range(epochs):
        optimizer.zero_grad()  # setup 0 gradient
        outputs = model(x)     # model input
        loss = criterion(outputs, y)  # calculate loss
        
        # check NaN  
        if torch.isnan(loss).any():
            print(f"NaN loss detected at epoch {epoch}")
            # if NaN is exists then return the epoch
            return epoch  

        loss.backward()  # backward propagation
        optimizer.step()  # wpdate optimization
    # if not NaN then return -1
    return -1 
```

```python
# training the model
epoch_with_nan = train_model(x, y, model, criterion, optimizer)
print(f"Epoch with NaN loss: {epoch_with_nan}")
```
